This is very simple. We import YOLO, load a model, train the model on the dataset, show some results on an image, and export the model.

```python
from ultralytics import YOLO

# load model
model = YOLO("yolov8m.pt") 

# train
model.train(data="dataset.yaml", epochs=25, imgsz=416, plots=True) 
metrics = model.val() 
results = model("./measurements/m3/frames/frame_00001.png")
path = model.export(format="onnx")
```
